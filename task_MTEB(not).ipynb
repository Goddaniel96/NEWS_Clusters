```python
import requests
from bs4 import BeautifulSoup

# กำหนด URL ที่ต้องการดึงข้อมูล
url = "https://huggingface.co/spaces/mteb/leaderboard"

# ทำการร้องขอข้อมูลจากเว็บไซต์
response = requests.get(url)

# ใช้ BeautifulSoup เพื่อแปลงเนื้อหาที่ได้รับมาเป็น HTML
soup = BeautifulSoup(response.content, 'html.parser')

# ค้นหา span ที่มี class = 'svelte-1scun43'
span_data = soup.find_all('span', class_='token svelte-1scun43')

# แสดงผลข้อมูลที่ได้
for span in span_data:
    print(span.get_text())  # หรือสามารถใช้ span.text ได้เช่นกัน
#not use
```
